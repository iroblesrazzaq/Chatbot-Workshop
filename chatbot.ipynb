```python
import os
from google import generativeai as genai
from dotenv import load_dotenv
```

```python
# Load gemini api key from environmet
load_dotenv()
api_key = os.getenv("GEMINI_API_KEY")

genai.configure(api_key=api_key)
```

```python
# there are a lot of models we can choose from!
models = genai.list_models()
print("Available models:")
for model in models:
    if "generateContent" in model.supported_generation_methods:
        print(f"- {model.name}")
```

```python
model_name = "gemini-2.0-flash"  # This is a basic model, you can change this

# Create the model object
model = genai.GenerativeModel(
    model_name=model_name,
    # You can adjust these parameters if needed
    generation_config={
        "temperature": 0.7,  # Controls randomness (0.0 to 1.0)
        "max_output_tokens": 500,  # Maximum response length
    }
)
```

```python



instructions = """You are a helpful and friendly AI assistant.
                 Respond in a casual and conversational tone.
                 Keep answers concise but informative."""


instructions = "You are a chatbot that responds like you are Skepta"

def ask_chatbot(query, instructions):
    """
    Send a query to the chatbot and get a response.
    
    Parameters:
    - query (str): The question or prompt for the chatbot
    - persona (str): The personality/style for responses ("default", "shakespeare", "pirate")
    
    Returns:
    - The chatbot's response text
    """


    # Format the complete prompt with instructions and user query
    full_prompt = f"{instructions}\n\nUser query: {query}"
    # call to model occurs here
    response = model.generate_content(full_prompt)
    return response.text
```
